```python
from sentence_transformers import SentenceTransformer, util

MODEL_NAME = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
model = SentenceTransformer(MODEL_NAME, device="cpu")
print(model.device)
```

## To-Do 1 · Crea las frases

Copia exactamente estos textos en A, B y C; conserva su orden en `sentences`:

- A = "Quiero devolver el producto que compré"
- B = "Necesito hacer una devolución de mi compra"
- C = "¿Qué tiempo hará este fin de semana?"

```python
A = "Quiero devolver el producto que compré"
B = "Necesito hacer una devolución de mi compra"
C = "¿Qué tiempo hará este fin de semana?"
sentences = [A, B, C]
```

## To-Do 2 · Genera los embeddings

Completa la llamada a `model.encode` con las frases. Conserva `convert_to_tensor=True`, igual que en el live coding, y muestra `embeddings.shape`. ¿Cuántas frases y cuántos valores por frase hay?

<details>
<summary>Pista 1 · Entrada del modelo</summary>

Recuerda que `model.encode` puede recibir una lista de frases. Ya has creado esa lista en `sentences`.

</details>

```python
embeddings = model.encode(sentences, convert_to_tensor=True)
print(embeddings.shape)
```

La forma esperada para este modelo es `torch.Size([3, 384])`: tres filas, una por frase, y 384 valores por fila. Comprueba la salida real; no necesitas interpretar cada componente del vector.

## To-Do 3 · Calcula las similitudes

Calcula A↔B y A↔C con `util.cos_sim`. Extrae cada número con `.item()` y usa los `print` preparados para mostrar cuatro decimales. La similitud no es un porcentaje ni una probabilidad.

<details>
<summary>Pista 2 · Selección de frases</summary>

Para comparar A y B necesitas `embeddings[0]` y `embeddings[1]`. C corresponde a `embeddings[2]`.

</details>

<details>
<summary>Pista 3 · Extraer el número</summary>

`util.cos_sim(...)` recibe los dos vectores y devuelve un tensor con un único valor; `.item()` permite extraerlo. El formato `:.4f` muestra cuatro decimales.

</details>

```python
similarity_ab = util.cos_sim(embeddings[0], embeddings[1]).item()
similarity_ac = util.cos_sim(embeddings[0], embeddings[2]).item()

print(f"A ↔ B: {similarity_ab:.4f}")
print(f"A ↔ C: {similarity_ac:.4f}")
```

**Resultado observado durante la validación en Google Colab:** A ↔ B = `0.7669`; A ↔ C = `-0.0830`. Son los valores de esa ejecución concreta, no constantes universales ni resultados obligatorios. Lo relevante es comparar ambas puntuaciones.
